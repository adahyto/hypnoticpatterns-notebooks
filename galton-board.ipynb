At every peg a ball bounces left or right at random. One ball tells you nothing, but thousands pile up into the same bell shape every time, because there are many more ways to land in the middle than at the edges.

## The script

Run the cell below (Shift + Enter). numpy and matplotlib are already in Colab.

```python
"""Galton board: every ball goes left or right at random at each of 12 rows of pegs.
Run: python galton-board.py   (needs: pip install numpy matplotlib)
"""
from math import comb

import numpy as np
import matplotlib.pyplot as plt

plt.style.use("dark_background")

rows, balls, p = 12, 5000, 0.5  # p: the chance of bouncing right
rng = np.random.default_rng()

bins = (rng.random((balls, rows)) < p).sum(axis=1)  # a ball's bin = how often it went right
counts = np.bincount(bins, minlength=rows + 1)

k = np.arange(rows + 1)
expected = [balls * comb(rows, i) * p**i * (1 - p) ** (rows - i) for i in k]  # the binomial curve

plt.figure(figsize=(7, 4))
plt.bar(k, counts, color="#5eead4", label="balls")
plt.xlabel("bin = times the ball went right")
plt.plot(k, expected, color="white", label="C(n, k) p^k (1-p)^(n-k)")
plt.legend()
plt.show()

# Try: p = 0.65 tilts the board; rows = 24 makes the bell narrower for its width
```

## Line by line

1. `rng = np.random.default_rng()`  
   numpy’s random number generator. Give it a number, default_rng(1), and every run gives the same “random” numbers.
2. `bins = (rng.random((balls, rows)) < p).sum(axis=1)`  
   A table of 5,000 × 12 random numbers between 0 and 1. Each one below p is True, a bounce to the right; adding up a row counts its Trues, which is the ball’s bin.
3. `counts = np.bincount(bins, minlength=rows + 1)`  
   How many balls landed in each bin from 0 to 12; minlength makes sure the empty edge bins are counted as 0.
4. `expected = [balls * comb(rows, i) * p**i * (1 - p) ** (rows - i) for i in k]`  
   A list comprehension: the binomial formula for every bin i, times the number of balls.
5. `plt.bar(k, counts, color="#5eead4", label="balls")`  
   One bar per bin. The white line drawn after it is the formula, to compare.

## Why it works

With 12 rows, a ball’s bin is simply the number of times it went right. There is one route to the far-left bin (left twelve times) but 924 routes to the centre (any six lefts and six rights), out of 4,096 routes in all.

The number of routes to bin k is the binomial coefficient C(12, k), the twelfth row of Pascal’s triangle: 1, 12, 66, 220, 495, 792, 924… More routes, more balls.

If each bounce goes right with probability p, the chance of bin k is C(n, k) · pᵏ · (1 − p)ⁿ⁻ᵏ: the binomial distribution. Its centre is n·p, and its spread grows only like √(n·p·(1 − p)), so more rows make a relatively narrower bell.

The de Moivre–Laplace theorem, a special case of the central limit theorem, says that with many rows this shape approaches the normal curve. Tilt the board in the demo: the bell moves, but it does not break. Francis Galton first showed his board in 1874 and wrote about it in Natural Inheritance (1889).

`P(k) = C(n, k) · pᵏ · (1 − p)ⁿ⁻ᵏ      centre = n·p      spread = √(n·p·(1 − p))`
