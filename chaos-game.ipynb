Pick one of a triangle’s three corners at random and jump halfway towards it. Mark the spot and repeat. The jumps are pure chance, yet after a few thousand of them the Sierpiński triangle appears: a triangle with triangles cut out of it, at every scale.

## The script

Run the cell below (Shift + Enter). numpy and matplotlib are already in Colab.

```python
"""Chaos game: jump a part r of the way to a random corner of a polygon, again and again.
Run: python chaos-game.py   (needs: pip install numpy matplotlib)
"""
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("dark_background")

corners, r, no_repeat = 3, 0.5, False  # triangle, halfway: the Sierpinski triangle
steps = 60000
rng = np.random.default_rng()

angles = np.pi / 2 + 2 * np.pi * np.arange(corners) / corners
vertex = np.column_stack([np.cos(angles), np.sin(angles)])

points, chosen = np.zeros((steps, 2)), np.zeros(steps, dtype=int)
x, last = np.array([0.0, 0.0]), -1
for i in range(steps):
    c = rng.integers(corners)
    while no_repeat and c == last:  # the restricted game: never the same corner twice
        c = rng.integers(corners)
    x = x + r * (vertex[c] - x)  # the whole rule
    points[i], chosen[i], last = x, c, c

plt.figure(figsize=(6, 6))
plt.scatter(points[20:, 0], points[20:, 1], s=0.3, c=chosen[20:], cmap="rainbow")
plt.axis("equal")
plt.axis("off")
plt.show()

# Try: corners, r = 5, 0.618 (a pentagon) or corners, no_repeat = 4, True (a square, restricted)
```

## Line by line

1. `vertex = np.column_stack([np.cos(angles), np.sin(angles)])`  
   The corners of the polygon: their x and y values side by side, one row per corner.
2. `c = rng.integers(corners)`  
   A random whole number from 0 to corners − 1: the corner to jump towards.
3. `while no_repeat and c == last:`  
   The restricted game: as long as the new corner is the last one, draw again.
4. `x = x + r * (vertex[c] - x)`  
   The rule: move the part r of the way from the point to the corner. With r = 0.5, halfway.
5. `points[i], chosen[i], last = x, c, c`  
   Keep the point and its corner, and remember the corner for the next step.
6. `plt.scatter(points[20:, 0], points[20:, 1], s=0.3, c=chosen[20:], cmap="rainbow")`  
   Skip the first 20 points, which are still on their way to the fractal, and colour the rest by their corner.

## Why it works

Jumping halfway to a corner halves every distance. Whatever the points fill, each of the three jumps maps it onto a half-size copy next to its corner. The Sierpiński triangle is made of exactly three half-size copies of itself, so it is the one shape the jumps leave unchanged, and every point is pulled onto it within a few jumps.

The order of the corners does not matter, only that every corner keeps coming up. That is why chance draws such a precise picture.

The three jumps form an iterated function system, and the triangle is its attractor. Change the rule and the attractor changes. With a square and jumps of one half the points just fill the square, but forbid the same corner twice in a row and a fractal appears. A pentagon needs a jump of about 0.618 of the way, the golden ratio again, for its five copies to just touch; a hexagon needs two thirds.

Colour each point by the corner it jumped towards, as the film and the demo do, and you see the copies directly.

`xₖ₊₁ = xₖ + r · (corner − xₖ)      triangle: r = 1/2      pentagon: r ≈ 0.618      hexagon: r = 2/3`
