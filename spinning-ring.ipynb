Balls bounce under gravity inside a ring with a gap, and the ring spins. You might expect the spin to fling the balls around, but a frictionless spinning ring bounces them exactly like a still one. All the spin does is carry the gap past them.

## The script

Run the cell below (Shift + Enter). numpy and matplotlib are already in Colab.

```python
"""A ball under gravity in a ring with a gap that spins; the bounces are elastic and without friction.
Run: python spinning-ring.py   (needs: pip install numpy matplotlib)
"""
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("dark_background")

R, ball, g = 1.0, 0.05, 9.81  # ring radius (m), ball radius, gravity
gap, spin = np.radians(18), 2 * np.pi / 4  # an 18 degree gap, one turn every 4 seconds
p, v = np.array([0.3, 0.2]), np.array([1.8, 0.7])
dt, t, track = 0.0005, 0.0, []

while t < 60:
    v[1] -= g * dt
    p = p + v * dt
    t += dt
    track.append(p.copy())
    r = np.linalg.norm(p)
    if r > R - ball:  # the ball touches the wall, or is at the gap
        angle = np.arctan2(p[1], p[0])
        centre = -np.pi / 2 + spin * t  # where the gap is now
        off = abs((angle - centre + np.pi) % (2 * np.pi) - np.pi)
        if off < gap / 2 - ball / R:  # all of the ball fits through: out
            break
        normal = p / r
        v = v - 2 * np.dot(v, normal) * normal  # the bounce: the spin plays no part in it
        p = normal * (R - ball)

track = np.array(track)
print(f"escaped after {t:.1f} s" if t < 60 else "still inside after 60 s")
plt.figure(figsize=(6, 6))
plt.plot(track[:, 0], track[:, 1], color="#5eead4", linewidth=0.4)
plt.gca().add_patch(plt.Circle((0, 0), R, fill=False, color="white"))
plt.axis("equal")
plt.axis("off")
plt.show()

# Try: spin = 0 and centre = -np.pi / 2 (a still gap at the bottom): the bounces stay the same
```

## Line by line

1. `v[1] -= g * dt`  
   Gravity: every step, the vertical speed drops by g·dt.
2. `p = p + v * dt`  
   The ball moves by its velocity times the time step. This simplest way to step is called Euler’s method.
3. `if r > R - ball:`  
   The ball’s edge reaches the wall when its centre is one ball radius away from the ring.
4. `off = abs((angle - centre + np.pi) % (2 * np.pi) - np.pi)`  
   The angle between the ball and the middle of the gap, wrapped into −π … π, so that 359° counts as 1° away.
5. `v = v - 2 * np.dot(v, normal) * normal`  
   The reflection: remove the part of the velocity along the normal twice. The spin of the ring does not enter it.
6. `p = normal * (R - ball)`  
   Put the ball back on the wall, so that a step can never leave it outside.

## Why it works

When a ball hits the wall, only the part of its velocity pointing into the wall is reversed; the part along the wall is kept. A spinning ring’s wall only moves along itself, sideways to the ball, so it has nothing to push the ball in or out with. Without friction the bounce is the same as off a still wall.

The bounces are perfectly elastic, so no energy is lost and the balls never settle. Gravity keeps them low, so their chance comes about once per turn, when the gap sweeps past the bottom.

A ball escapes only if all of it fits through the gap as it arrives: its centre must cross the opening at least one radius away from both edges. That is why the film widens the gap each round, from 8° to 12° and then 18°.

Without gravity a circle is a circular billiard, one of the few whose motion is exactly predictable: every bounce leaves at the angle it came in, so a ball traces chords of the same length for ever. Gravity bends the paths and breaks that order, so which ball escapes first depends on fine details of the start; the film’s run was picked from 500 simulated starts.

`bounce: v → v − 2·(v · n)·n      the wall moves along t ⊥ n, so (v · n) does not care about the spin`
