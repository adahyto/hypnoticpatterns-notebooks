## The script

Run the cell below (Shift + Enter). numpy and matplotlib are already in Colab.

```python
"""Times table on a circle: N dots, each dot n joined to dot (k * n) mod N.
Run: python times-table.py   (needs: pip install numpy matplotlib)
"""
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("dark_background")

N, k = 200, 2  # the number of dots and the multiplier

start = 2 * np.pi * np.arange(N) / N  # dot n at the angle 2*pi*n/N
end = start * k  # dot k*n: angles wrap around the circle by themselves (that is the "mod N")

plt.figure(figsize=(6, 6))
for a, b in zip(start, end):
    plt.plot([np.cos(a), np.cos(b)], [np.sin(a), np.sin(b)], color="#f5c76b", linewidth=0.6)
plt.axis("equal")
plt.axis("off")
plt.show()

# Try: k = 3 for the nephroid, k = 101 to see it collapse into diameters
```

## Line by line

1. `N, k = 200, 2`  
   Two names at once: 200 dots and the multiplier 2. These are the two numbers to change.
2. `start = 2 * np.pi * np.arange(N) / N`  
   The angle of each dot: dot n sits at 2π·n / N, so the 200 dots share the circle evenly.
3. `end = start * k`  
   The angle of dot k·n. An angle past 2π simply goes round again, and that does the wrapping (k·n mod N) for us.
4. `for a, b in zip(start, end):`  
   zip pairs the two arrays: each time round, the loop gets one start angle with its end angle.
5. `plt.plot([np.cos(a), np.cos(b)], [np.sin(a), np.sin(b)], color="#f5c76b", linewidth=0.6)`  
   plt.plot takes the x values, then the y values, of the line’s two ends: one chord of the circle each time round.

## Why it works

The numbers wrap around the circle like hours on a clock: with 200 dots, 150 × 2 = 300 lands on dot 100, as 300 − 200 = 100. This is arithmetic modulo 200.

The lines are chords of the circle, and the curve you see is their envelope, the shape every line just touches. Times 2 gives a cardioid with one cusp, times 3 a nephroid with two cusps.

These curves are epicycloids: the path of a point on a circle rolling around a fixed circle. A whole multiplier k gives an epicycloid with k − 1 cusps, and the multipliers in between morph one into the next, as you can see by sliding k in the demo.

Some multipliers fall into much simpler figures. With 200 dots, times 101 sends every even dot to itself and every odd dot straight across the circle, leaving only diameters.

`dot n  →  dot (k · n) mod N      whole k  →  an epicycloid with k − 1 cusps`
