Each new seed turns 137.5° from the one before and sits a little further out. That angle never lines up with itself, so the seeds never form rows: they fill the flower evenly, with no gaps and no crowding.

## The script

Run the cell below (Shift + Enter). numpy and matplotlib are already in Colab.

```python
"""Golden angle: seed n at the angle n x 137.5 degrees and the distance sqrt(n) (Vogel, 1979).
Run: python golden-angle.py   (needs: pip install numpy matplotlib)
"""
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("dark_background")

n = np.arange(1, 1001)  # seed numbers
golden = np.pi * (3 - np.sqrt(5))  # 137.5 degrees in radians: 360 / phi**2
theta, r = n * golden, np.sqrt(n)  # the whole rule

x, y = r * np.cos(theta), r * np.sin(theta)
plt.figure(figsize=(6, 6))
plt.scatter(x, y, s=8, c=n, cmap="viridis")
plt.axis("equal")
plt.axis("off")
plt.show()

# Try: golden = np.radians(144) -> five straight spokes (144 degrees is 2/5 of a turn)
```

## Line by line

1. `n = np.arange(1, 1001)`  
   An array of the numbers 1 to 1000, one per seed. np.arange stops just before its end, so 1001 gives 1000.
2. `golden = np.pi * (3 - np.sqrt(5))`  
   The golden angle in radians, the unit numpy’s trigonometry uses: a full turn is 2π. π·(3 − √5) ≈ 2.39996 radians, which is 137.5°.
3. `theta, r = n * golden, np.sqrt(n)`  
   Arithmetic on an array works on every element: n * golden is 1000 angles at once, np.sqrt(n) 1000 distances. No loop is needed.
4. `x, y = r * np.cos(theta), r * np.sin(theta)`  
   From an angle and a distance (polar coordinates) to a point on the page: x = r·cos θ, y = r·sin θ.
5. `plt.scatter(x, y, s=8, c=n, cmap="viridis")`  
   One dot per seed, size 8, each coloured from the “viridis” colour map by its number n.
6. `plt.axis("equal")`  
   The same scale on both axes, so the flower comes out round, not squashed.

## Why it works

137.5° is the smaller part of a full turn split in the golden ratio φ ≈ 1.618: 360° ÷ φ² ≈ 137.5°.

A simple fraction of a turn makes the seeds line up. Half a turn gives two lines, two fifths (144°) five spokes, three eighths (135°) eight. Try them in the demo. The golden ratio is the number that fractions approximate worst, as its continued fraction is all ones, so the golden angle keeps missing every spoke.

The best fractions for φ are ratios of neighbouring Fibonacci numbers: 3/2, 5/3, 8/5, 13/8, 21/13, 34/21… Each near miss shows as a family of spirals, which is why a sunflower head shows, for example, 21 spirals one way and 34 the other.

Vogel’s model (1979) puts seed n at the angle n × 137.508° and the radius c·√n. The area of a disc grows with the square of its radius, so every seed gets the same room, and the seeds lie along a Fermat spiral.

`θₙ = n × 137.508°      rₙ = c · √n      137.508° = 360° ÷ φ²`
