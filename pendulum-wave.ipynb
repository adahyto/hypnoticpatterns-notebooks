## The script

Run the cell below (Shift + Enter). numpy and matplotlib are already in Colab.

```python
"""Pendulum wave: 24 pendulums making 51, 52, ... 74 swings in 60 seconds.
Run: python pendulum-wave.py   (needs: pip install numpy matplotlib)
"""
import numpy as np
import matplotlib.pyplot as plt

plt.style.use("dark_background")

g, cycle = 9.81, 60.0  # gravity in m/s^2, seconds until they are back in step
n = np.arange(51, 75)  # swings per cycle, one more for each pendulum
period = cycle / n
length = g * (period / (2 * np.pi)) ** 2  # small swings: T = 2 pi sqrt(L / g)
print("lengths in cm:", np.round(length * 100, 1))

fig, axes = plt.subplots(1, 4, figsize=(12, 4), sharey=True)
for ax, t in zip(axes, [0, 15, 20, 30]):
    swing = np.cos(2 * np.pi * t / period)  # where each bob is, from -1 to 1
    ax.scatter(swing, np.arange(len(n)), c=swing, cmap="cool", s=40)
    ax.set_title(f"t = {t} s")
    ax.set_xlim(-1.2, 1.2)
    ax.axis("off")
plt.show()

# Try: t = 60 s (or 75, 80, 90): the same pictures again, as every count is whole
```

## Line by line

1. `n = np.arange(51, 75)`  
   The 24 counts of swings per minute: 51, 52, … 74.
2. `period = cycle / n`  
   Each pendulum’s period: 60 seconds divided by its count.
3. `length = g * (period / (2 * np.pi)) ** 2`  
   The small-swing pendulum formula T = 2π√(L / g), solved for the length L.
4. `fig, axes = plt.subplots(1, 4, figsize=(12, 4), sharey=True)`  
   One figure with four plots side by side; the loop fills one plot per moment.
5. `swing = np.cos(2 * np.pi * t / period)`  
   Where each bob is at the time t, from −1 to 1: after t / period swings, the cosine of that many full turns.
6. `ax.set_title(f"t = {t} s")`  
   An f-string puts the value of t into the title.

## Why it works

The lengths are chosen so that in 60 seconds the longest pendulum makes 51 swings, the next 52, and so on, up to 74 for the twenty-fourth. Every count is a whole number, so at 60 s every pendulum is back where it started.

In between, neighbours differ by a fixed part of a swing. At 30 s that part is half a swing, so every other pendulum is on the opposite side; at 20 s the row splits into three groups, at 15 s into four. Move the time slider in the demo to these moments.

For small swings a pendulum’s period is T = 2π√(L/g), so the length for n swings a minute is L = g · (60 s ÷ 2πn)²: about 34 cm for 51 swings and 16 cm for 74. Nothing connects the pendulums; the travelling wave is only your eye joining up their phases.

Flaten and Parendo (2001) explained the patterns as aliasing: the row shows one smooth wave at 24 points only, much as a film camera makes a wheel seem to turn backwards. Ernst Mach built the first such demonstration around 1867.

`Tₙ = 60 s ÷ n      Lₙ = g · (Tₙ ÷ 2π)²      n = 51, 52, … 74`
